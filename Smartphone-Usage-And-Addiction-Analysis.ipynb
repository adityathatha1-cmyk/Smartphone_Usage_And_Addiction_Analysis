# WHAT IS SMARTPHONE ADDICTION?

Smartphone addiction, often referred to as "nomophobia" (fear of being without a mobile phone), is a behavioral condition characterized by excessive and uncontrolled use of smartphones that begins to interfere with daily life, mental health, sleep quality, and social relationships.
Unlike casual smartphone usage, addiction is marked by an inability to limit screen time, constant checking of notifications, prioritizing phone usage over real-world responsibilities, and experiencing anxiety or restlessness when the phone is unavailable.
Key indicators of smartphone addiction include:

Daily screen time exceeding healthy limits
Heavy dependence on social media platforms
Disrupted sleep patterns due to late-night phone usage
Declining academic or work performance
Elevated stress and anxiety levels linked to phone usage

In this dataset, addiction is classified into four levels — None, Mild, Moderate, and Severe — based on a combination of behavioral indicators such as daily screen time, social media hours, app opens per day, notifications per day, and their measurable impact on sleep and stress. Understanding these levels through data visualization will help us identify the key patterns and risk factors that drive smartphone addiction across different demographics.

## Understanding Addiction Levels

## Q1. What is the overall distribution of smartphone Addiction levels among users?

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Calculate percentages
pct = (df['addiction_level']
       .value_counts(normalize=True)
       .mul(100)
       .sort_index())

plt.figure(figsize=(8,5))

ax = sns.barplot(
    x=pct.index,
    y=pct.values
)

plt.title('Distribution of Smartphone Addiction Levels (%)')
plt.xlabel('Addiction Level')
plt.ylabel('Percentage (%)')

# Add percentage labels
for i, v in enumerate(pct.values):
    ax.text(i, v + 0.5, f'{v:.1f}%', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```

## Observation
- Moderate addiction category has the highest number of users.
- Not Addicted users form the smallest group.
- Severe addiction is also present among a significant number of users.

# Bivariate Analysis

## Q2. Which age groups are most vulnerable to Smartphone Addiction?

```python
df['Age_Group'] = pd.cut(
    df['age'],
    bins=[15,20,25,30,35,40],
    labels=['16-20','21-25','26-30','31-35','36-40']
)
```

```python
plt.figure(figsize=(10, 5))

ax = sns.countplot(
    x='Age_Group',
    hue='addiction_level',
    data=df
)

# This block reads 'ax' and automatically places the percentage labels
for c in ax.containers:
    labels = [f'{v.get_height() / len(df) * 100:.1f}%' if v.get_height() > 0 else '' for v in c]
    ax.bar_label(c, labels=labels, padding=3)

plt.title("Age Group vs Addiction Level")
plt.xlabel("Age Group")
plt.ylabel("Count")
plt.show()
```

## Observation

- Moderate addiction is the most common addiction level across all age groups.
- Severe addiction is also common among users aged 21–35.
- The 16–20 age group has comparatively fewer users than the other age groups.
- The highest counts of addicted users are observed in the 21–35 age range, suggesting that young adults are the most vulnerable to smartphone addiction.

## Q4. How do Screen time, Social media, and Gaming relate to addiction levels?

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Q4 — Daily Screen Time
avg_sum = df.groupby('addiction_level')['daily_screen_time_hours'].mean().sum()
sns.barplot(x='addiction_level', y='daily_screen_time_hours', data=df, ax=axes[0])
for c in axes[0].containers:
    axes[0].bar_label(c, labels=[f'{v.get_height()/avg_sum*100:.1f}%' for v in c], padding=3)
axes[0].set_title('Daily Screen Time vs Addiction Level')
axes[0].set_xlabel('Addiction Level')
axes[0].set_ylabel('Daily Screen Time (Hours)')

# Q5 — Social Media Hours
avg_sum = df.groupby('addiction_level')['social_media_hours'].mean().sum()
sns.barplot(x='addiction_level', y='social_media_hours', data=df, ax=axes[1])
for c in axes[1].containers:
    axes[1].bar_label(c, labels=[f'{v.get_height()/avg_sum*100:.1f}%' for v in c], padding=3)
axes[1].set_title('Social Media Usage vs Addiction Level')
axes[1].set_xlabel('Addiction Level')
axes[1].set_ylabel('Social Media Hours')

# Q6 — Gaming Hours
avg_sum = df.groupby('addiction_level')['gaming_hours'].mean().sum()
sns.barplot(x='addiction_level', y='gaming_hours', data=df, ax=axes[2])
for c in axes[2].containers:
    axes[2].bar_label(c, labels=[f'{v.get_height()/avg_sum*100:.1f}%' for v in c], padding=3)
axes[2].set_title('Gaming Hours vs Addiction Level')
axes[2].set_xlabel('Addiction Level')
axes[2].set_ylabel('Gaming Hours')

plt.suptitle('Usage Drivers vs Addiction Level', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

## Observations
### Daily Screen Time:

- Screen time rises steadily with addiction — from 16.7% (Not Addicted) all the way to 31.9% (Severe).
- Severe users spend nearly 2x the screen time of non-addicted users.

### Social Media Usage:

- Not Addicted and Mild users are very close — 18.8% vs 19.0%, meaning mild addiction doesn't change social media habits much.
- A sharp jump happens at Moderate (30.1%) and Severe (32.2%), suggesting heavy social media use is a strong marker of deeper addiction.

### Gaming Hours:

- Gaming shows the steepest climb — from 15.6% (Not Addicted) to 34.3% (Severe).
- The jump from Mild (21.8%) to Moderate (28.3%) is notably large, indicating gaming escalates quickly once addiction sets in.

### Overall:

- All three usage drivers — screen time, social media, and gaming — consistently increase with addiction severity.
- Gaming shows the widest gap between non-addicted and severe users, making it the strongest behavioral indicator of smartphone addiction among the three.

## Q6. Do frequent Notifications contribute to Smartphone Addiction?

```python
plt.figure(figsize=(10, 6))

# 1. Draw the violin plot
sns.violinplot(
    x='addiction_level',
    y='notifications_per_day',
    data=df,
    inner='box' # Keeps the mini boxplot inside
)

# 2. Overlay a point plot to highlight and label the average notifications
ax = sns.pointplot(
    x='addiction_level',
    y='notifications_per_day',
    data=df,
    estimator='mean',
    color='white',     # White dot for contrast
    errorbar=None
)

# 3. Print the actual average text next to each point
means = df.groupby('addiction_level')['notifications_per_day'].mean()
for i, v in enumerate(means):
    ax.text(i, v + 5, f'Avg: {v:.1f}', color='black', weight='bold', ha='center')

plt.title('Notifications per Day vs Addiction Level (with Averages)')
plt.xlabel('Addiction Level')
plt.ylabel('Notifications Per Day')
plt.show()
```

## Observation
• The average number of notifications received per day increases as smartphone addiction severity increases.

• Users with Severe addiction receive the highest number of notifications per day, while Not Addicted users receive the lowest.

• Mild and Moderate addiction groups show similar notification levels, but both receive more notifications than Not Addicted users.

• The overall upward trend suggests that frequent notifications are associated with higher smartphone engagement and addiction levels.

## Q7. How does Smartphone Addiction affect Sleep duration?

```python
avg_sum = df.groupby('addiction_level')['sleep_hours'].mean().sum()

ax = sns.barplot(x='addiction_level', y='sleep_hours', data=df)

for c in ax.containers:
    labels = [f'{v.get_height()/avg_sum*100:.1f}%' for v in c]
    ax.bar_label(c, labels=labels, padding=3)

plt.title('Sleep Hours vs Addiction Level')
plt.xlabel('Addiction Level')
plt.ylabel('Sleep Hours')
plt.show()
```

## Observation
- Average sleep duration decreases consistently as addiction severity increases.

- Not Addicted users report the highest average sleep duration (6.7 hours), while Severe addiction users report the lowest (4.5 hours).


- The decline in sleep hours from Not Addicted to Severe users suggests a strong negative relationship between smartphone addiction and sleep duration.

- Users with higher addiction levels tend to sacrifice sleep for increased smartphone usage.

## Q9. What are the key behavioral differences between Addicted and Non-Addicted users?

```python
comparison = df.groupby('addicted_label')[
    ['daily_screen_time_hours',
     'social_media_hours',
     'gaming_hours',
     'sleep_hours']
].mean()
import matplotlib.pyplot as plt
import numpy as np

# Columns: daily_screen_time_hours, social_media_hours, gaming_hours, sleep_hours

fig, ax = plt.subplots(figsize=(10, 6))

comparison.T.plot(kind='bar', ax=ax)

# Calculate percentage difference
for i, metric in enumerate(comparison.columns):

    non_addicted = comparison.loc[0, metric]
    addicted = comparison.loc[1, metric]

    pct_change = ((addicted - non_addicted) / non_addicted) * 100

    # Position label above the highest bar
    y_pos = max(non_addicted, addicted) + 0.3

    ax.text(
        i,
        y_pos,
        f'{pct_change:.1f}%',
        ha='center',
        fontsize=11,
        fontweight='bold'
    )

# Add value labels on bars
for container in ax.containers:
    ax.bar_label(container, fmt='%.1f')

ax.set_title('Behavioral Differences Between Addicted and Non-Addicted Users')
ax.set_ylabel('Average Hours')
ax.set_xlabel('Usage Metrics')
ax.legend(['Non-Addicted', 'Addicted'])
plt.xticks(rotation=25)

plt.tight_layout()
plt.show()
```

## Observations
- Addicted users spend 64.2% more time on their phones daily (8.5 vs 5.2 hours) compared to non-addicted users.
- Social media usage is 64.1% higher among addicted users (3.7 vs 2.3 hours).
- Addicted users spend 59.5% more time gaming (4.0 vs 2.5 hours).
- Non-addicted users get 21.5% more sleep (6.2 vs 4.8 hours), indicating a negative impact of excessive smartphone use on sleep duration.
- The largest behavioral differences are observed in screen time, social media usage, and gaming activity, making them strong indicators of smartphone addiction.

## Q10. Which behavioral factors are most strongly associated with Smartphone Usage Patterns?

```python
plt.figure(figsize=(10,6))

corr_cols = [
    'daily_screen_time_hours',
    'social_media_hours',
    'gaming_hours',
    'work_study_hours',
    'sleep_hours',
    'notifications_per_day',
    'app_opens_per_day',
    'weekend_screen_time'
]

sns.heatmap(
    df[corr_cols].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Smartphone Usage Variables')
plt.show()
```

## Heatmap Observations

1. Daily screen time and weekend screen time show a very strong positive correlation (0.94).

- Users who spend more time on smartphones during weekdays also tend to spend more time on weekends.

2. Sleep hours have a negative correlation with screen time.

- Higher smartphone usage is associated with lower sleep duration.

3. Gaming hours show a moderate positive correlation with screen time and weekend usage.

- Users who spend more time gaming generally have higher overall smartphone usage.

4. Work/study hours have little to no correlation with other variables.

- Smartphone usage patterns appear to be influenced more by entertainment activities than work or study.

## Summary Insight
- Smartphone usage is mainly driven by gaming, notifications, and frequent app interactions, while excessive screen time is associated with lower sleep duration, highlighting potential signs of smartphone dependency.
